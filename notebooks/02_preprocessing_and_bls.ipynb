# 02 — Preprocessing and BLS

Create one reproducible NPZ sample per TOI. Catalogue ephemerides are retained only for diagnostic recovery checks, not passed to BLS or the model.

```python
from pathlib import Path
from transit_hunter.data import build_sample_from_arrays, read_sample
from transit_hunter.download import load_tic_arrays_with_sectors

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
# Replace with one TIC from the frozen, eligible pilot table.
TOI, TIC = '1064.01', 79748331
time, flux, sectors = load_tic_arrays_with_sectors(TIC, ROOT / 'data/raw')
sample_path, metadata = build_sample_from_arrays(toi=TOI, time=time, flux=flux, sector_labels=sectors, output_dir=ROOT / 'data/processed')
metadata['candidate']
```

```python
sample = read_sample(sample_path)
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(10, 3))
axes[0].plot(sample['global_view'], lw=1); axes[0].set_title('Global view')
axes[1].plot(sample['local_view'], lw=1); axes[1].set_title('Local view')
sample['diagnostics']
```
